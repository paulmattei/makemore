```python
# Provides data about which char comes after which char. But there is more data here, which char comes first, and which last.
for w in words[:1]:
    for ch1, ch2 in zip(w, w[1:]):
        print(ch1, ch2)
```

```python
# Add a mechanism to define what the start and end char are
for w in words[:1]:    
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        print(ch1, ch2)
```

```python
# We want to be able to count ocurances of bigrams. To do this we add a dictionary to keep track.
b = {}
for w in words:    
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        bigram = (ch1, ch2)
        b[bigram] = b.get(bigram, 0) + 1
print(len(b))
```

```python
sorted(b.items(), key=lambda kv: kv[1], reverse=True)[:10]
```

```python
# It will be convenient for us to store these results in an array, rather than a python dict.
import torch
```

```python
N = torch.zeros((28,28), dtype=torch.int32) # 28 = 26 letters of alphabet + <'S'> and <'E'>
```

```python
# Get a list of all unique chars
chars = sorted(list(set(''.join(words)))) # set() will dedup as memebers of set() must be unique
# String to Int
stoi = {s:i for i,s in enumerate(chars)}
stoi['<S>'] = 26
stoi['<E>'] = 27
stoi
```

```python
for w in words:    
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1
```
